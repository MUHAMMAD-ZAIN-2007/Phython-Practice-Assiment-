## A food delivery order has 3 items, each costing Rs. 450. The delivery fee is Rs. 100
    Create variables and calculate the final bill.

```python
item1 = "Zinger_Burger"
item2 = "Chicken_Biryani"
item3 = "Clab_Sandwitch"
price = 450
delivery = 100
result = price * 3 
final_bill = result + delivery

print("========== FOOD DELIVERY BILL ==========")
print(f"Item 1: {item1} - Rs. {price}")
print(f"Item 2: {item2} - Rs. {price}")
print(f"Item 3: {item3} - Rs. {price}")
print("----------------------------------------")
print(f"Items Total: Rs. {result}")
print(f"Delivery Fee: Rs. {delivery}")
print("----------------------------------------")
print(f"Final Bill: Rs. {final_bill}")
print("========================================")
```
